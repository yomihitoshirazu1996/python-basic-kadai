```python
class Human:

    def __init__(self, name, age):
        self.name = name
        self.age = age

    def check_adult(self):
        if self.age >= 20:
            print(f"{self.name}は大人である")
        else:
            print(f"{self.name}は大人でない")

human1 = Human("侍太郎", 36)
human2 = Human("花子", 18)
human3 = Human("ジョン", 25)
human4 = Human("メアリー", 10)

human_list = [human1, human2, human3, human4]

for human in human_list:
    human.check_adult()
```
